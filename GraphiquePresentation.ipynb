**Création de tableaux de valeurs** (à adapter) :

```python
x = linspace(0, 10, 24)    # 24 valeurs régulièrement espacées de 0 à 10
y = sin(x) + 0.2 * random(24)    # exemple : valeurs « mesurées » proches de sin(x)
modele = sin(x)    # exemple : modèle théorique
```

**Tracé d'un graphique.** Chaque courbe est ajoutée avec `fig.add_trace(go.Scatter(...))` ; les choix (mode, légende, symboles, couleurs, incertitudes, titre, axes...) sont détaillés en commentaires de lignes. Le graphique s'affiche automatiquement en fin de cellule.

```python
fig = go.Figure()

fig.add_trace(go.Scatter(
    x=x, y=y,
    mode="markers",    # "markers" points, "lines" courbe, "lines+markers" les deux
    name="mesures",    # texte de la légende
    marker=dict(symbol="cross", size=9, color="blue")))    # ex. : symbol="+", size=12, color="red" ; autres couleurs : green, orange, purple, black...

fig.add_trace(go.Scatter(
    x=x, y=modele,
    mode="lines",
    name="modèle",
    line=dict(color="crimson", width=2, dash="dash")))    # - -- -. :

fig.update_layout(
    title="Mon titre",
    xaxis_title="temps (s)",
    yaxis_title="tension (V)",
    height=600)    # hauteur du graphique en pixels ; ex. : 450, 700...
```

**Plusieurs graphiques dans la même figure** avec `make_subplots(rows=..., cols=..., subplot_titles=(...))` :

```python
fig = make_subplots(rows=2, cols=1,
                    subplot_titles=("Graphique du haut", "Graphique du bas"))

fig.add_trace(go.Scatter(x=x, y=y, mode="markers", name="mesures"),
              row=1, col=1)    # emplacement du tracé
fig.add_trace(go.Scatter(x=x, y=modele, mode="lines", name="modèle"),
              row=2, col=1)

fig.update_xaxes(title_text="temps (s)", row=2, col=1)
fig.update_layout(height=650)
```

**Avec des incertitudes** (barres d'erreur) :

```python
fig = go.Figure()

fig.add_trace(go.Scatter(
    x=x, y=y,
    mode="markers",
    name="mesures",
    marker=dict(symbol="circle", size=8, color="blue"),
    error_y=dict(type="constant", value=0.15, color="black", width=3)))
    # incertitude constante ; sinon : error_y=dict(type="data", array=incertitudes)

fig.add_trace(go.Scatter(x=x, y=modele, mode="lines", name="modèle",
                         line=dict(color="crimson", width=2)))

fig.update_layout(title="Mesures avec incertitudes",
                  xaxis_title="temps (s)", yaxis_title="tension (V)",
                  height=600)
```

**Histogramme d'une série de mesures** (statistiques) :

```python
mesures = random(200)    # 200 valeurs aléatoires entre 0 et 1
fig = go.Figure()
fig.add_trace(go.Histogram(x=mesures, nbinsx=20, name="mesures",
                           opacity=0.85, marker_color="blue"))
fig.update_layout(title="Histogramme", xaxis_title="valeur",
                  yaxis_title="effectif", height=600)
```

**Repère orthonormé** (même échelle en x et en y) :

```python
t = linspace(0, 2 * pi, 200)
fig = go.Figure()
fig.add_trace(go.Scatter(x=cos(t), y=sin(t), mode="lines"))
fig.update_yaxes(scaleanchor="x", scaleratio=1)    # repère orthonormé
fig.update_layout(title="Cercle trigonométrique",
                  xaxis_title="x", yaxis_title="y", height=600)
```

**Échelle logarithmique** :

```python
x2 = linspace(1, 1000, 50)
fig = go.Figure()
fig.add_trace(go.Scatter(x=x2, y=x2**2, mode="markers", name="y = x²"))
fig.update_yaxes(type="log")    # échelle logarithmique sur l'axe des ordonnées
fig.update_layout(title="Échelle logarithmique",
                  xaxis_title="x", yaxis_title="y", height=600)
```

**Vecteurs** (avec `fig.add_annotation`) :

```python
fig = go.Figure()
for xi, yi, dx, dy in zip(array([0, 1, 2]), array([0, 0, 0]),
                          array([1, 1, 1]), array([1, 0, -1])):
    fig.add_annotation(x=xi + dx, y=yi + dy, ax=xi, ay=yi,
                       axref="x", ayref="y", showarrow=True,
                       arrowhead=2, arrowwidth=2, arrowcolor="red")
fig.update_yaxes(scaleanchor="x", scaleratio=1)
fig.update_layout(title="Vecteurs", xaxis_title="x", yaxis_title="y",
                  height=600)
```
